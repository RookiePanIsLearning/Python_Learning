# 模組 8：API 串接與資料解析

目標：使用 requests 發送 GET 請求並解析 JSON。

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts/1"
response = requests.get(url, timeout=10)
response.raise_for_status()
data = response.json()

print("status:", response.status_code)
print("title:", data.get("title"))
```
