# 模組 5：檔案讀寫 (File I/O)

目標：練習使用 with open 讀寫文字檔。

```python
from pathlib import Path

demo_file = Path("sample_note.txt")

with open(demo_file, "w", encoding="utf-8") as f:
    f.write("Python File I/O demo\n")
    f.write("line 2\n")

with open(demo_file, "r", encoding="utf-8") as f:
    content = f.read()

print(content)
```
