# 模組 2：條件判斷 (if/elif/else)

目標：實作基礎邏輯判斷。

```python
score = 78

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "D"

print(f"score={score}, grade={grade}")
```

```python
# 練習：輸入年齡並判斷是否成年
age = 17
is_adult = age >= 18
print(f"age={age}, is_adult={is_adult}")
```
