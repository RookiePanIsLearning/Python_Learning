```python
# === 範例 1：解包 (Unpacking) ===
# 基本解包
point = (10, 20)
x, y = point
print(f"x={x}, y={y}")

# 星號解包
first, *middle, last = [1, 2, 3, 4, 5]
print(f"first={first}, middle={middle}, last={last}")

# 變數交換
a, b = 1, 2
a, b = b, a
print(f"交換後: a={a}, b={b}")
```

```python
# === 範例 2：運算子 ↔ 特殊方法 ===
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(1, 2)
v2 = Vector(3, 4)
v3 = v1 + v2           # 語法糖，實際呼叫 v1.__add__(v2)
print(v3)               # Vector(4, 6)
print(v1 == v2)         # False（呼叫 v1.__eq__(v2)）
```

```python
# === 範例 3：鏈式比較 ===
score = 75
if 60 <= score < 80:
    print("及格")       # 等同 (60 <= score) and (score < 80)
```

```python
# === 範例 4：海象運算子 := ===
# 在推導式中一行搞定計算 + 過濾
data = [5, 12, 3, 18, 7, 2]
filtered = [y for x in data if (y := x * 3) > 10]
print(f"海象過濾: {filtered}")   # [36, 54, 21]
```

```python
# === 範例 5：f-string 進階技巧 ===
import math
pi = math.pi
print(f"圓周率: {pi:.4f}")           # 3.1416
print(f"{'Python':>15}")              # 右對齊
print(f"{'Python':-^20}")             # 居中填充
print(f"{pi = }")                     # debug 用法: pi = 3.14...
```

```python
# === 範例 6：match-case 模式匹配 (Python 3.10+) ===
def classify(value):
    match value:
        case int(n) if n > 0:
            return f"正整數: {n}"
        case int(n):
            return f"非正整數: {n}"
        case str(s):
            return f"字串: {s}"
        case [x, y]:
            return f"二元素列表: [{x}, {y}]"
        case _:
            return "其他"

print(classify(42))
print(classify(-3))
print(classify("hello"))
print(classify([1, 2]))
```

```python
# --- 練習 8：綜合挑戰 — 用語法糖重構囉嗦程式碼 ---
# 需求：把下面的「囉嗦版」重構成 Pythonic 版（盡量用語法糖）

data = {"users": [
    {"name": "Pan", "scores": [85, 92, 78]},
    {"name": "Alice", "scores": [90, 88, 95]},
    {"name": "Bob", "scores": [70, 65, 80]},
]}

# 囉嗦版：找出平均分數 > 80 的使用者名稱
result_verbose = []
for user in data["users"]:
    total = 0
    for s in user["scores"]:
        total += s
    avg = total / len(user["scores"])
    if avg > 80:
        result_verbose.append(user["name"])

print(result_verbose)   # ['Pan', 'Alice']

# Pythonic 版（用一行 list comprehension + sum() 重寫）
# result_pythonic = ???
```
